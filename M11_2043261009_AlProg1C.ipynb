```python
def cek_genap(angka):
  if angka%2==0:
    return True
  else:
    return False
status_angka=cek_genap(10)
print(f'Status angka 10: {status_angka}')
```

```python
def hitung_luas_persegi(sisi):
  luas=sisi*sisi
  return luas

luas_persegi1=hitung_luas_persegi(5)
print(f'Luas persegi adalah: {luas_persegi1}')
luas_persegi2=hitung_luas_persegi(10)
print(f'Luas persegi adalah: {luas_persegi2}')
luas_persegi3=hitung_luas_persegi(15)
print(f'Luas persegi adalah: {luas_persegi3}')


rata_rata=(luas_persegi1+luas_persegi2+luas_persegi3)/3
print(f'rata-rata luas ketiga persegi tersebut adalah {rata_rata}')
```

```python
def konversi_nilai(nilai_angka):
  if nilai_angka >= 86:
    return "A"
  elif nilai_angka >= 76 and nilai_angka < 86:
    return "AB"
  elif nilai_angka >= 66 and nilai_angka < 76:
    return "B"
  else:
    return "C"

nilai=int(input('Masukkan nilai anda:'))
indeks=konversi_nilai(nilai)
print(f'Nilai huruf anda adalah: {indeks}')
```
